```python
import pandas as pd
```

```python
path = "ventas.xlsx"

df_ventas = pd.read_excel(path)
display(df_ventas.info())
```

Obsevaciones: 
- Se tienen que eliminar columnas que no aporten datos de valor
  - id_venta (sirve para relacionarlo a las otras tablas)
  - id_cliente (seria dato personal y no es relevante)
  - nombre_cliente (mismo que arriba)
  - email (idem)
- La fecha ya esta en datetime, solo quedaria convertir el medio_pago en string

```python
display(df_ventas.describe())
```

```python
df_clean = df_ventas.drop(columns=["id_venta", "id_cliente", "nombre_cliente", "email"])
df_clean["fecha"] = df_clean["fecha"].replace(" ","")
display(df_clean.describe())
```

Agrego columnas para tener mas data asociada a fecha
- Mes
- Cuatrimestre
- Semana

```python
df_clean["fecha"] = pd.to_datetime(df_clean['fecha'])
df_clean["mes"] = df_clean["fecha"].dt.to_period("M")

display(df_clean.describe(include="all"))
```

```python
# Agrego cuatrimestre y trimestre
df_clean["cuatrimestre"] = ((df_clean["fecha"].dt.month - 1) // 4 + 1).astype(int)
df_clean["anio_cuatrimestre"] = df_clean["fecha"].dt.year.astype(str) + "-C" + df_clean["cuatrimestre"].astype(str)

display(df_clean.describe(include="all"))
display(df_clean)
```
